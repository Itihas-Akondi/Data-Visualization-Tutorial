# Fan Plot Data Visualization Tutorial

## Introduction

This notebook is a step-by-step guide to understanding and implementing a fan plot in your visualizations. A fan plot enhances a regular line plot by introducing new dimensions of data representations. With a fan plot, one can visualize the spread/variation, or uncertainty of the dataset's values. In this example, I will use rolling percentiles to show how we can use a fan plot to smooth out a trend over time, separate noise from the dataset's general trend, and display the variance with the help of the shaded regions. This tutorial focuses on the smoothing and variation within the dataset, as well as how these features can be represented through sensible design choices.

As discussed in Data Feminism by Catherine D'Ignazio and Lauren F. Klein, visualizations like a fan plot can help "visceralize" data by making patterns feel more meaningful and easier to interpret, as the shaded regions in this type of plot can signify the variance in the measurement or the uncertainty, depending on the analysis and the dataset. 



## Why a Fan Plot?


The reason I chose the fan plot is that Matplotlib's documentation mainly focuses on building different types of bar plots, line plots, and scatterplots, where visualizations like the fan plot are usually not well-known, even when they show the variation, uncertainty, or smooth out our trends in our data over a period of time for our line plot. The simple, effective nature, along with the lack of documentation in matplotlib, motivates me to create this tutorial that guides users through a step-by-step process to build their own fan plot.

Also, according to Cairo, a visualization like the fan plot is highly valuable for a project, as it clearly shows data patterns without oversimplifying. It also aligns with Edward Tufte's teachings, which state that a good plot conveys rich information efficiently, much like a fan plot.

````python
import matplotlib.pyplot as plt 
import pandas as pd 
import numpy as np
````

## Tutorial 

###   First we will import and clean the dataset that we will be using for this tutorial

````python
dat_fan = pd.read_csv("GLB.Ts+dSST.csv",skiprows=1)
#display(dat_fan)
dat_fan.columns
````

````python
dat_fan = dat_fan[["Year", "J-D"]]
dat_fan_2 = dat_fan.dropna()
dat_fan_2 = dat_fan_2.rename(columns={"J-D": "Annual_Anomaly_C"})
dat_fan_2["Annual_Anomaly_C"] = pd.to_numeric(dat_fan_2["Annual_Anomaly_C"], errors="coerce")
display(dat_fan_2)
````

### From here we will build the base line plot of our data that we will show the variation in through the fan plot

````python

plt.figure(figsize=(12, 6))
plt.plot(dat_fan_2["Year"],dat_fan_2["Annual_Anomaly_C"])
plt.title("Global Annual Temperature Anomaly")
plt.xlabel("Year")
plt.ylabel("Temperature anomaly (°C)")
plt.grid()
````

### After making your base line plot, we will now try to find the median and the percentile values to make the ranges of our line plot around the central median.

The function that we will be using to do this is `pandas.DataFrame.rolling` function that allows us to provide the window of calculation, in this case 30 years. This means that for each year, Python looks at a 30-year range of temperature anomaly values and calculates different percentiles within that window.

The 50th percentile here will be the rolling median, the 25th-75th and the 10th-90th percentile will show the variation in the line plot and the spread of uncertainty.

````python

dat_fan_2["p10"] = dat_fan_2["Annual_Anomaly_C"].rolling(window=30, center=True, min_periods=1).quantile(0.10)
dat_fan_2["p25"] = dat_fan_2["Annual_Anomaly_C"].rolling(window=30, center=True, min_periods=1).quantile(0.25)
dat_fan_2["p50"] = dat_fan_2["Annual_Anomaly_C"].rolling(window=30, center=True, min_periods=1).quantile(0.50)
dat_fan_2["p75"] = dat_fan_2["Annual_Anomaly_C"].rolling(window=30, center=True, min_periods=1).quantile(0.75)
dat_fan_2["p90"] = dat_fan_2["Annual_Anomaly_C"].rolling(window=30, center=True, min_periods=1).quantile(0.90)

display(dat_fan_2)
````

### Plotting our visualization



Now with the required data, in order to make our fan plot, we will utilize a mtplotlib function called `fill_between()`. This fuction will take in the inputs of lower and upper bounds along the x axis and shade a band between these two bounds.

```python
plt.fill_between(x, lower_bound, upper_bound)
```

This will give us our distinct fan plot look and help us study the uncertainty in our measurement of the line plot trend.



````python
plt.figure(figsize=(12,6))

plt.fill_between(dat_fan_2["Year"],dat_fan_2["p10"],dat_fan_2["p90"],alpha=0.18,color="tab:blue",label="10th–90th percentile range")
plt.fill_between(dat_fan_2["Year"],dat_fan_2["p25"],dat_fan_2["p75"],alpha=0.35,color="tab:blue",label="25th–75th percentile range")


plt.plot(dat_fan_2["Year"],dat_fan_2["p50"],linewidth=2.5,color="tab:blue",label="30 year rolling median")
plt.plot(dat_fan_2["Year"],dat_fan_2["Annual_Anomaly_C"],color="Black",linewidth=1,alpha=0.5,label="Annual Anomaly in (°C)")
plt.title("Fan Plot of Global Annual Temperature Anomaly")
plt.xlabel("Year")
plt.ylabel("Temperature anomaly (°C)")
plt.legend()
plt.grid()
````

## Design Choices 

To properly convey the idea behind your visualizations while avoiding misleading the audience's thought process, making the right color and design choices in your fan plot is crucial. An example of this is the fan plot visualization presented above. Here, I have used the color blue to tie together the three related statistical measures in our dataset. The varying transparency of the percentile ranges here indicates the narrowing of the bands toward the actual rolling median, with darker shades indicating the inner percentile range closest to it. The distinct yet not overpowering black line here distinguishes it from the other three features, as it shows the raw annual anomaly dataset. 

When making a fan plot, the user should focus on using varying transparency and color changes to represent the intensity features of similar statistical measures and to differentiate between those that differ, using contrasting colors. In the end, the user should focus on how their work, at first glance, will subconsciously shape the public's view and how that affects their main goal.
